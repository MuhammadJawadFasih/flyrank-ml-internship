```python
from datetime import date

train = dataset["train"]

def find_first_date(target):
    low = 0
    high = len(train) - 1

    while low <= high:
        mid = (low + high) // 2
        if train[mid]["report_date"] < target:
            low = mid + 1
        else:
            high = mid - 1
    return low

march_start = find_first_date(date(2026, 3, 1))
march_end = find_first_date(date(2026, 4, 1))

print("March start:", march_start)
print("March end:", march_end)
print("Rows in March:", march_end - march_start)
```

## 1. Unit of analysis + time window

*One row = one what, over which dates? State it, then verify it below.*

1. Unit of analysis + time window

My lane is Refresh / Content Opportunity Scoring.

One row represents one content page with its historical search and engagement signals.

The time window used for verification is a mid-panel month: 2026-03. This avoids using the final month as development data and helps prevent label leakage.

## 2. Fields: feature / label / context / excluded

*Sort every field you plan to touch into these four buckets. Excluded needs a why.*



### Features

- gsc_impressions: Search impressions measured before the decision moment.
- gsc_clicks: Search clicks measured from previous search activity.
- gsc_avg_position: Average search ranking position available from previous performance.
- ga4_sessions: User sessions measured from analytics history.
- scroll_events: User engagement events available from previous activity.

These features are available before making a content refresh decision.

### Label

Refresh opportunity score (proxy): A future performance-based score or ranking target created from historical outcomes.

### Context

- client_hash_id: Identifies the client.
- content_hash_id: Identifies the content page.
- report_date: Provides the time context.

### Excluded

- Future performance after the decision date.
- Any information generated after the prediction moment.

These are excluded because they would leak future information and create unrealistic model performance.

## Leakage experiment

To demonstrate leakage, a proxy outcome column is created from engagement performance. This outcome information would normally only be available after the decision moment.

The leaked version of this outcome is intentionally added as a feature. It is removed afterward because it would make evaluation unrealistic.

```python
# Small leakage demonstration

leak_demo = march_2026.select(range(100))

leak_demo = leak_demo.add_column(
    "leaked_label",
    leak_demo["gsc_clicks"]
)

print(leak_demo)
```

## Leakage removal

The leaked_label column is removed because it contains outcome information that would not be available at the decision moment. Keeping it would create an unrealistic model evaluation.

## 4. Data limits

*What can this data never tell you? Unbalanced history, GSC-only early rows, window overlaps.*


This dataset has some limitations.

- It cannot tell the actual quality of content or whether a human editor would consider a page worth refreshing.
- It only contains measured search and engagement signals, so some important factors outside the data are missing.
- Some rows may have missing Search Console or Google Analytics availability.
- The analysis uses March 2026 as a development window, so the findings may not represent all time periods or future changes.

Therefore, the results should be treated as directional and decision-support rather than absolute conclusions.
